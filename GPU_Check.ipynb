```python
import os
import logging
import warnings

# 1. Silence most TensorFlow C++ logs (INFO / WARNING)
# 0 = all logs, 1 = filter INFO, 2 = filter WARNING, 3 = filter ERROR
os.environ["TF_CPP_MIN_LOG_LEVE"] = "1"

# 2. Standard Python warnings
warnings.filterwarnings("ignore")
```

```python
import tensorflow as tf
tf.get_logger().setLevel(logging.ERROR)
from tensorflow.python.client import device_lib
```

```python
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print("GPU Available: Yes")
    print("GPU Name: ", device_lib.list_local_devices()[-1].physical_device_desc)
else:
    print("GPU Available: No")
```
