#### Data Extraction Function 

```python
import numpy as np
import pandas as pd
import zipfile
import urllib.request
```

```python
url = 'https://github.com/mattharrison/datasets/raw/master/data/kaggle-survey-2018.zip'
fname = 'kaggle-survey-2018.zip'
member_name = 'multipleChoiceResponses.csv'
```

```python
def extract_zip(src,dst,member_name):
    """Extract a memeber file from a zip file and read it into a panadas dataframe
    Parameter:
        src(str): URL of the target zip file
        dst (str): Local file path where the zip file will be written
        member_name(str): name of the memeber file inside the zip file to pull data from
    
    Returns: 
        a Panadas DataFrame containing the conentsd of the member_name file
    """

    url = src
    fname = dst
    fin = urllib.request.urlopen(url)
    data = fin.read()

    with open(dst, mode='wb') as fout:
        fout.write(data)
    with zipfile.ZipFile(dst) as z:
        kag = pd.read_csv(z.open(member_name))
        kag_questions = kag.iloc[0]
        raw = kag.iloc[1:]
        return raw
```

```python
raw = extract_zip(url, fname, member_name)
```
