```python
import torch
```

```python
print("PyTorch version:", torch.__version__)
```

```python
print("Is CUDA available?", torch.cuda.is_available())
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Current device:", device)
```

```python
x = torch.rand(5, 3)
y = torch.rand(5, 3)
z = x + y
print("Tensor addition result:\n", z)
```

```python
# Verify computation device for tensor
print("Tensor device:", z.device)
```

```python
x = torch.randn(1000, 1000, device="cpu")
y = torch.matmul(x, x)
print("Matrix multiplication successful on:", y.device)
```
